```python
# Contoh array multidimensional (Nesting) dalam bentuk matriks
M = [ [2,3,4],
      [5,6,7],
      [8,9,0] ]

print('Matriks 3x3')
print(M)
      
print('Mengambil baris ke-2')
print(M[1])

print('Mengambil baris ke-2, Lalu dapatkan item ke-3 didalam baris')
print(M[1][2])
print(M[0][2])
print(M[2][2])
Matrix = ((M[1][2]), (M[0][2]), (M[2][2]))
print(Matrix)
```

```python
# Membuat Matriks 
M = [ [2,3,4],
      [5,6,7],
      [8,9,0] ]

print('Matriks 3x3')
print(M)

# Tambahkan 1 item untuk setiap item dikolom 2
M1 = [row[1]+1 for row in M]
print('Hasil Penambahan 1 Untuk Setiap Item dIkolom 2')
print(M1)

# Menyaring item pada matriks
print('Matriks 3x3 sebelum penyaringan data :')
print(M)
M2 = [row[1] for row in M if row[1] %2 == 0]
print('Data  matriks yang telah disaring pada kolom ke-2')
print(M2)

M3 = [row[0] for row in M if row[0] %2 == 0]
print('Data matriks yang telah disaring pada kolom ke-1')
print(M3)

# Mengumpulkan diagonal dari matriks
print('Matriks 3x3 :')
print(M)
diag = [M[i][i] for i in [0,1,2]]
print('Bentuk diagonal matriks')
print(diag)
```

```python
#  Menambahkan elemen kedalam list
buah = ['Banana','バナナ','Pisang']
print('Element Awal :')
print(buah)

#Menggunakan metode append (); untuk menambah elemen tunggal dibagian akhir list
buah.append('banán')
print('\nsetelah append()')
print(buah)

#Menggunakan metode insert(); untuk menambahkan elelmen tunggal di indeks/posisi tertentu 
buah.insert(4,'바나나')
print('\nsetelah insert :')
print(buah)

#Menggunakan metode extend(); untuk menyambung atau menggabungkan suatu list dengan yang lainnya 
buah.extend(['香蕉','Banane'])
print('\nsetelah extend :')
print(buah)

```

```python
# Menghapus elemen dari list 

# Membuat List 
barang = ['palu','obeng','paku','pisau','peniti','baut']
print('Sebelum Dihapus')
print(barang)

# Menggunakan remove(); menghapus elemen list berdasarkan nilainya 
# Bentuk umum : nama_variable.remove()
barang.remove('palu')
barang.remove('peniti')
print('setelah dihapus dengan metode remove :')
print(barang)

# Menggunakan dell[]; menghapus elemen list berdasarkan indeksnya 
# Bentuk umum : del nama_variable_[indeks]
del barang[2]
del barang [-1]
print('setelah dihapus dengan metode del :')
print(barang)


# Menggunakan clear[]; menghapus elemen didalam list
print(len(barang))
barang.clear()
print(barang)
```

 # TYPE TUPLE 

```python
# Type tuple adalah list yang tidak dapat dirubah 
# Tuple sebenarnya mirip dengan array konstan 
# Bentuk umum tuple menggunakan tanda kurung: Nama_tuple = (nilai1,nilai2,nilai3...) 

bulan = ('januari','februari','maret','april','mei','juni','juli')
print(type(bulan))

# Menampilkan Bulan 
print(bulan)
print(bulan[0])
print(bulan[3])
print(bulan[2])
print(bulan[4])
```

```python
# Contoh operasi pada tuple 
T = ( 10,20,30,40) 
print(T)

print(len(T))
print(T + (50,60))
print(T[0])
print(T.index(30))
print(T.count(40))

T1 = ('Bilangan',4.0, [100,200,300])
print(T1)
print(T1[1])
print(T1[2][1])
```

```python
# Bentuk umum tuple menggunakan tanpa tanda kurung : nama_tuple = nilai1,nilai2,nilai3,...
bulan = ('januari','februari','maret','april','mei','juni','juli')
print(type(bulan))

# Menampilkan bulan 
print(bulan)
print(bulan[1])
print(bulan[3])
print(bulan[2])
```

```python
# Tuple didefinisikan menggunakan tanda kurung(), bukan bracket[]
# Tuple yang dibuat tanpa tanda kurung perlu menggunakan tanda backlash (\) untuk menyambung baris perintah pertama dan kedua 
# Beberapa hal yang perlu diperhatikan mengenai tuple 


bulan = ('januari','februari','maret','april','mei','juni','juli')
print('Data Tuple')
print(bulan)

# Tuple diindeks menggunnakan bilangan dan diambil dari 0 
print('indeks positif :')
print(bulan[4])

# Indeks negatif pada tuple dapat digunakan dan -1 selalu menunjukkan elemen terakhir 
print('indeks negatif :')
print(bulan[-4])
```

```python
# Tuple tidak memiliki metode append()
bulan = ('januari','februari','maret','april','mei','juni','juli')
print('Data tuple awal')
print(bulan)
bulan.append('Bulan Libur')
print('setelah append():')
print(bulan)
```

```python
# Elemen Tuple tidak dapat diubah 
bulan = ('januari','februari','maret','april','mei','juni','juli')
print('Data Tuple Awal')
print(bulan)

# Mengubah nilai elemen tuplle 
bulan[3] = 'Desember'
# Menampilkan hasil perubahan 
print('setelah diubah :')
print(bulan)
```

```python
# Tuple tidak memiliki metode Remove()
bulan = ('januari','februari','maret','april','mei','juni','juli')
print('Data Tuple Awal')
print(bulan)

# Mengubah nilai elemen tuplle 
bulan.remove('Desember')
# Menampilkan hasil perubahan 
print('setelah diremove :')
print(bulan)
```

```python
# Tuple memeliki metode indeks()
bulan = ('januari','februari','maret','april','mei','juni','juli')
print('Data Tuple Awal')
print(bulan)

print(bulan[0])
indeks = bulan.index('januari') 
print(indeks)
```

```python
# Tuple menggunakan operator "in" digunakan untuk memeriksa apakah termasuk item pada tuple atau tidak 
bulan = ('januari','februari','maret','april','mei','juni','juli')
print('Data Tuple Awal')
print(bulan)

print('februari pada item Tuple')
print('februari' in bulan)
print('desember pada item Tuple')
print('desember' in bulan)
```
