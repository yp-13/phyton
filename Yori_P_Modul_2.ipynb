# MODUL DUA

```python
# Type String 
print('Ohayou Gozaimasu')
print('Ohayou Gozaimasu')
```

```python
# Variabel Type String 
s1 = 'Ohayou'
s2 = 'Gozaimasu'

# Menampilkan Type String
print(s1)
print(s2)
print(s1+''+s2)
```

```python
# Membandingkan String 
s1 = 'Ohayou'
s2 = 'Gozaimasu'

print(s1 > s2)
print(s1 >= s2)
print(s1 < s2)
print(s1 <= s2)
print(s1 != s2)
```

```python
# Mengambil Substring; Menggunakan tanda braket []
s = 'Konichiwa'

print(s[0])
print(s[-3])
print(s[:2])
print(s[2:])
print(s[-3:])
print(s[0:1])
len(s)
print(s +' '+ 'Minnasan')
print(s * 2)
```

```python
# Menentukan Offset Substring 
s = 'Konichiwa'
s.find('me')
```

```python
# Menggatikan Kemunculan Suatu Substring Dengan Substring Lainnya 
s = 'Konichiwa'
s.replace('me', 'ME')
```

```python
# Mengkonversikan Huruf Besar dan Kecil 
s = 'Konichiwa'
s.upper()
```

# TYPE NUMERIK 

```python
# Type Integer 
a = 3
b = -19
c = 200

print(type(a))
print(type(b))
print(type(c))
```

```python
# Type Float 
x = 0.19
y = -2000
z = 30.9

print(type(x))
print(type(y))
print(type(z))
```

```python
# Type Komplek
s1 = 3 + 3j
s2 = -78.5j

print(type(s1))
print(type(s2))
print(s1.real) #Menampilkan Bilangan Real
print(s1.imag) #Menampilkan Bilangan Imajiner
```

```python
# Konversi Tipe Data Numerik
# Konversi ke Integer 
x = int (0.19)
print(x)

# Konversi Ke float  
b = (-2000)
print(b)

# Konversi ke Complex
y = complex(30.9)
print(y)
```

```python
# Operasi Aritmatika 
a = 25
b = 9
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)

```

```python
# Modul Math
import math 
a = 4 
b = 5
c = 0
print(math.sqrt(a))
print(math.pi*(b))
print(math.cos(c))
```

```python
# Modul Math 
import cmath
z = 1 + 5j
print(cmath.sqrt(z)) #akar variable z
print(cmath.sin(z)) #akar variable z
```
