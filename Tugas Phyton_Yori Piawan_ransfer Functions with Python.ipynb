# Time Constant T

```python
import numpy as np
import control
import matplotlib.pyplot as plt

# Define constants
K = 10
Tarray = [5, 10, 30, 50]
start = 0
stop = 200
step = 0.1
t = np.arange(start, stop, step)

# Initialize the plot
plt.figure()

# Generate step responses for each T
for T in Tarray:
    # Create Transfer Function
    num = np.array([K])
    den = np.array([T, 1])
    H = control.tf(num, den)
    print('H(s) =', H)

    # Step Response
    t, y = control.step_response(H, t)

    # Plot
    plt.plot(t, y, label=f'T = {T}')

# Set up the plot
plt.title("Step Response for different T")
plt.xlabel("Time (s)")
plt.ylabel("Response (y)")
plt.legend()
plt.grid()
plt.show()
```

# Pade Approx

```python
import numpy as np
import matplotlib.pyplot as plt
import control

# Time Delay
Tau = 2
N_values = [1, 2, 3, 5, 10]  # Corrected N to N_values
start = 0
stop = 10
step = 0.1
t = np.arange(start, stop, step)

# Initialize plot
plt.figure()

# Generate step responses for each N
for N in N_values:
    # Pade Approximation
    num_pade, den_pade = control.pade(Tau, N)
    Hpade = control.tf(num_pade, den_pade)
    print(f'Hpade(s) for N={N} =', Hpade)

    # Step Response
    t, y = control.step_response(Hpade, t)

    # Plot each step response with the label
    plt.plot(t, y, label=f'N={N}')  # Added label for legend

# Set up the plot
plt.title("Pade Approximations for Different N")
plt.xlabel("Time (t)")
plt.ylabel("Response y(t)")
plt.legend()  # Show legend with N values
plt.grid(True)
plt.show()
```

# Orde With Time Delay

```python
import numpy as np
import matplotlib.pyplot as plt
import control
K = 3
T = 4
num = np.array ([K])
den = np.array ([T , 1])
H1 = control.tf(num , den)
print ('H1(s) =', H1)
Tau = 2
N = 5 # Order of the Approximation
[num_pade,den_pade] = control.pade(Tau,N)
Hpade = control.tf(num_pade,den_pade)
print ('Hpade(s) =', Hpade)
H = control.series(H1, Hpade)
print ('H(s) =', H)
t, y = control.step_response(H)
plt.plot(t,y)
plt.title("H(s)")
plt.grid()
```

# Poles and Zeros

```python
import numpy as np
import control
import matplotlib.pyplot as plt

# Define initial transfer function H(s)
num = np.array([2, 3, 1])  # Numerator coefficients
den = np.array([12, 7, 1])  # Denominator coefficients
H = control.tf(num, den)
print("Initial H(s):")
print(H)

# Now you want to combine two transfer functions using convolution

# First transfer function H1
num1 = np.array([1, 1])  # Numerator for H1
den1 = np.array([3, 1])  # Denominator for H1
H1 = control.tf(num1, den1)

# Second transfer function H2 (Corrected: passing both numerator and denominator)
num2 = np.array([2, 1])  # Numerator for H2
den2 = np.array([4, 1])  # Denominator for H2
H2 = control.tf(num2, den2)  # Both num2 and den2 are passed here

# Combine H1 and H2 in series
H_series = control.series(H1, H2)
print("Series connection H(s) =")
print(H_series)

# Get and print the zeros and poles of the combined transfer function
zeros = control.zeros(H_series)  # Corrected: use control.zeros() instead of control.zero()
print("Zeros of H(s):", zeros)

poles = control.poles(H_series)  # Corrected: use control.poles() instead of control.pole()
print("Poles of H(s):", poles)

# Plot the pole-zero map for the series connection
control.pzmap(H_series)
plt.show()
```

# Pole and Stablity

```python
import control
import numpy as np
import matplotlib.pyplot as plt

# Define Transfer Function
num = np.array([2])  # Numerator
den = np.array([3, 1])  # Denominator
H = control.tf(num, den)
print("H(s) =", H)

# Poles of the Transfer Function
p = control.poles(H)
print("Poles of H(s):", p)

# Step Response
t, y = control.step_response(H)

# Plot the Step Response
plt.figure(figsize=(8, 4))  # Create a new figure for step response
plt.plot(t, y, label="Step Response")
plt.title("Step Response")
plt.xlabel("Time (s)")
plt.ylabel("Response y(t)")
plt.grid(True)
plt.legend()

# Plot the Pole-Zero Map
plt.figure(figsize=(8, 4))  # Create a new figure for the pole-zero map
control.pzmap(H, plot=True, grid=True)  # Ensure pole-zero map is plotted
plt.title("Pole-Zero Map")  # Corrected string

# Show both plots
plt.show()
```

# Marginally Stable System

```python
import control
import numpy as np
import matplotlib.pyplot as plt

# Define Transfer Function
num = np.array([3])  # Numerator
den = np.array([1, 0, 1])  # Denominator (s^2 + 1)
H = control.tf(num, den)
print("H(s) =", H)

# Poles of the Transfer Function
p = control.poles(H)  # Use control.poles for correct pole computation
print("Poles of H(s):", p)

# Step Response
tstart = 0
tstop = 20
tstep = 0.1
t = np.arange(tstart, tstop, tstep)
t, y = control.step_response(H, t)

# Plot Step Response
plt.figure(figsize=(8, 4))
plt.plot(t, y, label="Step Response")
plt.title("Step Response")
plt.xlabel("Time (s)")
plt.ylabel("Response y(t)")
plt.grid(True)
plt.legend()

# Pole-Zero Map
plt.figure(figsize=(8, 4))
control.pzmap(H, plot=True, grid=True)
plt.title("Pole-Zero Map")

# Show plots
plt.show()

# Analyze Stability
if all(np.real(pole) < 0 for pole in p):
    print("Stable System: All poles have negative real parts.")
elif any(np.real(pole) > 0 for pole in p):
    print("Unstable System: At least one pole has a positive real part.")
else:
    print("Marginally Stable System: Poles lie on the imaginary axis without any positive real parts.")
```
